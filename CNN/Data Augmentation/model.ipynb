```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img = image.load_img('train/cat1.jpg', target_size=(200,200))
```

```python
img
```

```python
type(img)
```

```python
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)
```

```python
img = image.img_to_array(img)
```

```python
img.shape
```

```python
input_batch = img.reshape(1,200,200,3)
# 1 image of (200,200,3) dimension
```

```python
# Creates 10 images and then stops
i=0
for output in datagen.flow(input_batch,batch_size=1,save_to_dir='aug'):
  i=i+1
  if i==10:
    break

# flow -> is used when there is only 1 image to transform
# flow_from_directory -> is used when there is a folder full images divided into classes.
```
