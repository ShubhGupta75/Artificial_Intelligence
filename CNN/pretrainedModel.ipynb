```python
from tensorflow.keras.applications.resnet50 import ResNet50
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.resnet50 import preprocess_input, decode_predictions
import numpy as np
```

```python
model=ResNet50(weights='imagenet')
```

```python
# Path of the image that we want to predict
img_path = '/content/dog2.webp'

# Load the image and resize it to 224 × 224 pixels
# ResNet50 expects an input image of this size
img = image.load_img(img_path, target_size=(224, 224))

# Convert the PIL image into a NumPy array
# Shape becomes (224, 224, 3) for an RGB image
x = image.img_to_array(img)

# Add a batch dimension because the model expects a batch of images
# Shape changes from (224, 224, 3) → (1, 224, 224, 3)
x = np.expand_dims(x, axis=0)

# Apply ResNet50's ImageNet-specific preprocessing
# This prepares the pixel values in the format expected by the pretrained model
x = preprocess_input(x)
```

```python
img
```

```python
preds = model.predict(x)
print("Predicted: ", decode_predictions(preds, top=3)[0])
```
