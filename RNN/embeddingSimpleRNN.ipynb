```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.models import Sequential
from keras.layers import Embedding,Dense,Input
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
model = Sequential()

model.add(Input(shape=(5,)))

model.add(Embedding(
    input_dim=18,
    output_dim=2
))

model.summary()

# vocab length = 17, output_dim = dimension of dense vector we want.
# input_dim = len(tokenizer.word_index) + 1 = vocab + 1
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

# Sentiment Analysis

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
# Method to find vocab_size
vocab_size = max(max(sequence) for sequence in X_train) + 1 # +1 because embedding indices start at 0.
vocab_size
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()

model.add(Input(shape=(50,)))
model.add(Embedding(input_dim=vocab_size, output_dim=2)) # Embedding layer, limiting input_dim to 10000
model.add(SimpleRNN(32,return_sequences=False)) # RNN layer
model.add(Dense(1, activation='sigmoid')) # output layer

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
